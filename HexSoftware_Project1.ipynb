# Key insight and Conclusion

# 1.Dataset Overview
* 150 flowers total (50 of each species: Setosa, Versicolor, Virginica)
* Measures 4 features: sepal.length/width, petal.length/width in cm

# 2.Key Patterns
* Versicolor: mid-range between the other two and petal features show clear species seperation than sepal features.
* Virginica: Largest measurements overall, the petals are greater than 5cm
* Setosa: Distinctly smaller petals that is the length is less than 2cm and width less than 0.7cm

# 3.Strong Correlations
* Petal length and width: 0.96 which is very strong
* Sepal length and petal length: 0.87
* Sepal width shows weak correlations with other features

# 4.Practical Implications 
* Petal measurements are better for species identification
* Setosa can be identified by its uniquely small petals
* Virginica stands out with largets dimension
